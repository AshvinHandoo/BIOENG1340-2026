# Gyromagnetic Ratio of a Proton from the Larmor Equation

The **Larmor equation** relates the precession (resonance) frequency of a spin to the static
magnetic field:
$$\omega_{\text{Larmor}} = \gamma\, B_0,$$
where
- $\omega_{\text{Larmor}}$ is the **angular** resonance frequency in rad/s,
- $\gamma$ is the **gyromagnetic ratio** (rad s$^{-1}$ T$^{-1}$),
- $B_0$ is the static field in tesla (T).

In terms of ordinary (cyclic) frequency $f$ in hertz, $\omega = 2\pi f$, so
$$f_{\text{Larmor}} = \frac{\gamma}{2\pi}\, B_0.$$

**Given (from the figure):**
- $B_0 = 2.35\ \text{T}$
- resonance frequency $f_{\text{Larmor}} = 100\ \text{MHz}$ (a 2.35 T magnet is the classic
  "100 MHz" proton NMR system)

We solve for $\gamma$, being careful with **unit conversions** (MHz $\to$ Hz, and cyclic $\to$
angular frequency).

```python
import numpy as np
```

## Inputs and unit conversions
The frequency is given in **MHz** and must be converted to **Hz**, then to **angular** frequency
($\omega = 2\pi f$) before applying $\omega = \gamma B_0$.

```python
# ---- Given data ----
B0_tesla   = 2.35        # static field [T]
f_MHz      = 100.0       # Larmor (cyclic) frequency [MHz]

# ---- Unit conversions ----
f_Hz    = f_MHz * 1e6            # MHz -> Hz
omega   = 2 * np.pi * f_Hz       # cyclic -> angular frequency [rad/s]

print(f"B0            = {B0_tesla} T")
print(f"f_Larmor      = {f_MHz} MHz = {f_Hz:.3e} Hz")
print(f"omega_Larmor  = 2*pi*f = {omega:.6e} rad/s")
```

## Solve for the gyromagnetic ratio
From $\omega = \gamma B_0$: $\ \gamma = \omega / B_0$. The reduced form $\gamma/2\pi$ (in MHz/T) is
the quantity usually quoted for protons.

```python
gamma        = omega / B0_tesla            # [rad s^-1 T^-1]
gamma_over_2pi_Hz = gamma / (2 * np.pi)    # [Hz/T]
gamma_over_2pi_MHz = gamma_over_2pi_Hz / 1e6

print(f"gamma           = {gamma:.6e} rad s^-1 T^-1")
print(f"gamma / (2*pi)  = {gamma_over_2pi_Hz:.6e} Hz/T")
print(f"gamma / (2*pi)  = {gamma_over_2pi_MHz:.4f} MHz/T")
```

## Compare with the accepted proton value
The CODATA proton gyromagnetic ratio is $\gamma_p = 2.6752218744\times10^{8}$ rad s$^{-1}$ T$^{-1}$,
i.e. $\gamma_p/2\pi = 42.577\ \text{MHz/T}$.

```python
gamma_accepted = 2.6752218744e8            # [rad s^-1 T^-1] (CODATA)
pct_error = 100 * abs(gamma - gamma_accepted) / gamma_accepted

print(f"computed  gamma = {gamma:.6e} rad s^-1 T^-1")
print(f"accepted  gamma = {gamma_accepted:.6e} rad s^-1 T^-1")
print(f"percent error   = {pct_error:.3f} %")
```
